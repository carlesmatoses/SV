## Exercise 1. 

Check the contents of the Vega dataset co2_concentration.json . Plot with the x
variable ‘Date’ as quantitative and y with variable ‘CO2’ as quantitative. What happens? Now
try to use the date as a temporal variable? What happens if you use the date as a categorical
variable?

```python
import pandas as pd
from vega_datasets import data
from vega_datasets import local_data
import altair as alt


print(data.list_datasets())
print(local_data.list_datasets())

# contents of co2-conccentration.json
df = data.co2_concentration()
print(df.head())
print(df.dtypes)

# add column 'DateQ' as quantitative
df['DateQ'] = pd.to_datetime(df['Date']).astype(int) / 10**9
print(df.head())

# add column 'DateS' as string
df['DateS'] = df['Date'].astype(str)
print(df.head()) 
```

```python
# Plot x=Date and y=CO2 both as quantitative using Altair
chart = alt.Chart(df).mark_point().encode(
    x='DateQ:Q',
    y='CO2:Q'
).properties(
    title='CO2 Concentration Over Time'
)
chart.show()
```

```python
# Plot x=Date and y=CO2 both as correct units
chart = alt.Chart(df).mark_point().encode(
    x='Date:T',
    y='CO2:Q'
).properties(
    title='CO2 Concentration Over Time'
)
chart.show()
```

```python
# Plot x=DateS and y=CO2 with date as string
chart = alt.Chart(df).mark_point().encode(
    x='DateS:N',
    y='CO2:Q'
).properties(
    title='CO2 Concentration Over Time'
)
chart.show()
```

## Exercise 4. 

With the previous example, now encode in the circle mark the population of the
country. Compare the sizes using circles, squares, and points.

```python
# Circles with size encoding for population
chart_circle = alt.Chart(df).mark_circle().encode(
    x='life_expect:Q',
    y='fertility:Q',
    size='pop:Q'
).properties(
    title='Life Expectancy vs Fertility (Circle, size=Population)'
)
chart_circle.show()

# Squares with size encoding for population
chart_square = alt.Chart(df).mark_square().encode(
    x='life_expect:Q',
    y='fertility:Q',
    size='pop:Q'
).properties(
    title='Life Expectancy vs Fertility (Square, size=Population)'
)
chart_square.show()

# Points with size encoding for population
chart_point = alt.Chart(df).mark_point().encode(
    x='life_expect:Q',
    y='fertility:Q',
    size='pop:Q'
).properties(
    title='Life Expectancy vs Fertility (Point, size=Population)'
)
chart_point.show()
```

# Exercises chapter 6.2

## Exercise 1. 

Use the Vega dataset wheat.json . Plot with the x variable ‘year’ as temporal and y
with variable ‘wheat’ as quantitative. Use a line plot, with a dashed pink line of width 3.

```python
df = data.wheat()
df.head()

chart = alt.Chart(df).mark_line(
    color='pink',
    strokeDash=[5, 5],
    strokeWidth=3
).encode(
    x='year:T',
    y='wheat:Q'
).properties(
    title='Wheat Production Over Time'
)
chart.show()
```

## Exercise 2. 

Use the same dataset to produce a scatterplot with gray triangles filled with blue.

```python
chart = alt.Chart(df).mark_point(
    shape='triangle-up',
    color='gray',
    fill='blue',
).encode(
    x='year:T',
    y='wheat:Q'
).properties(
    title='Wheat Production Over Time (Triangles)'
)
chart.show()
```

## Exercise 3. 

Increase the size of the previous triangles.

```python
chart = alt.Chart(df).mark_point(
    shape='triangle-up',
    color='gray',
    fill='blue',
    size=200
).encode(
    x='year:T',
    y='wheat:Q'
).properties(
    title='Wheat Production Over Time (Triangles)'
)
chart.show()
```

## Exercise 4. 

With the cars dataset, render a scatterplot of the acceleration versus the
horsepower, with the points encoded as red crosses with black stroke, and their size
proportional to the horsepower.

```python
df = data.cars()
df.head()
```

```python
chart = alt.Chart(df).mark_point(
    shape='cross',
    color='black',
    fill='red',
    stroke='black'
).encode(
    x='Horsepower:Q',
    y='Acceleration:Q',
    size='Horsepower:Q'
).properties(
    title='Acceleration vs Horsepower (Red Crosses, Size=Horsepower)'
)
chart.show()
```

## Exercise 5. 

Modify the previous plot so that the marks are circles with an opacity of 0.25 and
the outline is the same color than the filling color.

```python
chart = alt.Chart(df).mark_circle(
    opacity=0.25,
    color='red',
    stroke='red'
).encode(
    x='Horsepower:Q',
    y='Acceleration:Q',
    size='Horsepower:Q'
).properties(
    title='Acceleration vs Horsepower (Red Circles, Opacity 0.25, Size=Horsepower)'
)
chart.show()
```
